```python
import httpx
from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
http_client = httpx.Client(verify=False)
client = Anthropic(http_client=http_client)

model = "claude-sonnet-4-6"

def add_user_message(messages,text):
    user_message = {"role":"user","content":text}
    messages.append(user_message)

def add_assistant_message(messages,text):
    assistant_messages = {"role":"assistant","content":text}
    messages.append(assistant_messages)

def chat(messages):
    system = """You are patient math tutor
    Do not directly answer students questions
    Guide them to a solution step by step"""
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
        system=system
    )
    return message.content[0].text

messages =[]

add_user_message(messages,"How I solve 5x+3 =2 for x")

answer = chat(messages)

print(answer)
```
